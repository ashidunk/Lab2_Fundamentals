```python
# APPENDIX 1 - SENSOR READING MONITOR

ENGINEER = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
SEED_NUM = int(STUDENT_ID[-1])

# Set the number of sensor readings
NUMBER_OF_READINGS = SEED_NUM + 5

# Operating ranges
LOW_LIMIT = 30
HIGH_LIMIT = 70

sensor_readings = []
valid_readings = []
invalid_readings = []

print("=== SENSOR READING MONITOR ===")
print("Engineer:", ENGINEER)
print("Seed Number:", SEED_NUM)

# Generate sensor readings
for i in range(NUMBER_OF_READINGS):
    reading = (ord(ENGINEER[0]) + i * 10 + SEED_NUM) % 121
    sensor_readings.append(reading)

print("Generated Sensor Data:", sensor_readings)

# Classify readings
for reading in sensor_readings:

    if reading > 100:
        invalid_readings.append(reading)
        print(reading, "-> INVALID")

    else:
        valid_readings.append(reading)

        if reading < LOW_LIMIT:
            print(reading, "-> LOW")

        elif reading <= HIGH_LIMIT:
            print(reading, "-> NORMAL")

        else:
            print(reading, "-> HIGH")

# Summary
print("\n=== SUMMARY ===")
print("Valid Readings:", valid_readings)
print("Invalid Readings:", invalid_readings)
print("Number of Valid Readings:", len(valid_readings))
print("Number of Invalid Readings:", len(invalid_readings))
print("Execution Log: Sensor readings processed successfully.")
print("Final Output: Sensor monitoring completed.")
```

```python
# APPENDIX 2 - TEXT-BASED SIGNAL ANALYZER

ENGINEER = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
SEED_NUM = int(STUDENT_ID[-1])

# Generate signal
signal = ENGINEER + "-" + str(SEED_NUM)

letters = 0
digits = 0
spaces = 0
special = 0

print("=== SIGNAL ANALYZER ===")
print("Engineer:", ENGINEER)
print("Seed Number:", SEED_NUM)
print("Generated Signal:", signal)

# Process the signal character by character
for i in range(len(signal)):

    character = signal[i]

    if character.isalpha():
        print(character, "-> LETTER")
        letters += 1

    elif character.isdigit():
        print(character, "-> DIGIT")
        digits += 1

    elif character.isspace():
        print(character, "-> SPACE")
        spaces += 1

    else:
        print(character, "-> SPECIAL CHARACTER")
        special += 1

# Classify the signal
if letters > 0 and digits > 0:
    classification = "ALPHANUMERIC SIGNAL"

elif letters > 0:
    classification = "TEXT SIGNAL"

elif digits > 0:
    classification = "NUMERIC SIGNAL"

else:
    classification = "SPECIAL CHARACTER SIGNAL"

print("\n=== ANALYSIS ===")
print("Letters:", letters)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special Characters:", special)
print("Signal Classification:", classification)
print("Execution Log: Signal processed character by character.")
print("Final Output: Signal analysis completed.")
```

```python
# APPENDIX 3 - ELECTRONIC ACCESS CONTROL

ENGINEER = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
SEED_NUM = int(STUDENT_ID[-1])

# Generate password
generated_password = ENGINEER[:3].upper() + str(SEED_NUM) + ENGINEER[-2:].upper()

# Set attempt limit
ATTEMPT_LIMIT = SEED_NUM + 3

attempts = 0
access_granted = False

print("=== ELECTRONIC ACCESS CONTROL ===")
print("Engineer:", ENGINEER)
print("Seed Number:", SEED_NUM)
print("Generated Password:", generated_password)
print("Attempt Limit:", ATTEMPT_LIMIT)

# Analyze password character by character
print("\n=== PASSWORD ANALYSIS ===")

for i in range(len(generated_password)):

    character = generated_password[i]

    if character.isalpha():
        print(character, "-> LETTER")

    elif character.isdigit():
        print(character, "-> DIGIT")

    else:
        print(character, "-> SPECIAL CHARACTER")

# Password authentication
print("\n=== AUTHENTICATION ===")

for i in range(ATTEMPT_LIMIT):

    password = input("Enter Password: ")

    # Normalize input
    password = password.strip().upper()

    attempts += 1

    if password == generated_password:
        print("ACCESS GRANTED")
        access_granted = True
        break

    else:
        print("ACCESS DENIED")

# Final system state
print("\n=== SYSTEM STATUS ===")
print("Attempts Made:", attempts)

if access_granted:
    print("Access Result: GRANTED")
    print("Final System State: UNLOCKED")

else:
    print("Access Result: DENIED")
    print("Final System State: LOCKED")

print("Execution Log: Authentication process completed.")
print("Final Output: Access control completed.")
```
